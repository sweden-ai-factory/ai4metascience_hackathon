Use the CSC AITTA API to access LLMs hosted on the LUMI supercomputer.

1. Read the documentation:  
   [CSC AITTA API Documentation](https://docs.lumi-supercomputer.eu/laif/inference/aitta/)

2. Generate your API token:
Visit AITTA Portal and log in using your CSC, Haka, or MyAccessId/Puhuri account. Select your LUMI project and generate an access token.  
   [CSC AITTA Portal](https://aitta.csc.fi/)

3. Save your token in .env:

Open a JupyterLab terminal and run:

     vi .env

Add your token:

     API_KEY=your_aitta_token_here

Save and exit: press Esc, type :wq, then press Enter.

4. Access the token in Python:

```bash
from dotenv import load_dotenv
import os

load_dotenv()
API_KEY = os.getenv("API_KEY")
```

⚠️ Security: Never share your API token or commit .env to GitHub. Add .env to .gitignore.

````python
import os
from dotenv import load_dotenv
from openai import OpenAI
from langchain_openai import ChatOpenAI
````

````python
# Load environment variables from .env
load_dotenv()

# Read API key
API_KEY = os.getenv("API_KEY")

if not API_KEY:
    raise ValueError("API_KEY not found. Check your .env file.")

# Initialize OpenAI client
client = OpenAI(
    api_key=API_KEY,
    base_url="https://aitta-api.csc.fi/openai/v1"
)

# Call the model with streaming enabled
chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": "Write a novel about Quantum Computing."
        }
    ],
    model="openai/gpt-oss-120b",
    stream=True
)

# Print streamed response
for chunk in chat_completion:
    if chunk.choices and chunk.choices[0].delta.content is not None:
        print(
            chunk.choices[0].delta.content,
            end="",
            flush=True
        )
````
